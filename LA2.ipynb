```python
# Calculate the covariance matrix for TV, Radio, Newspaper and Sales.
target = ['TV', 'Radio','Newspaper','Sales']
cov_matrix = df[target].cov()
print(cov_matrix)
```

```python
# Calculate the Pearson correlation coefficient matrix.
rf = df[target].corr(method='pearson')
print(rf)
```

```python
df.plot(kind = 'scatter', x= 'TV',y='Sales')
```

```python
corr = df[target].corr()
sns.heatmap(corr,annot=True)
```
